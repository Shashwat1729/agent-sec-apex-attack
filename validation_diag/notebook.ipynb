```python
import time
from pathlib import Path

for i in range(18):
    hits = list(Path('/kaggle/input').rglob('*.gguf'))
    print(f'[{i}] t={i*10}s found {len(hits)} gguf files:', hits)
    if hits:
        break
    time.sleep(10)

print('\nFull /kaggle/input tree:')
for p in sorted(Path('/kaggle/input').rglob('*')):
    print(' ', p)
```
